```python
rating_counts = df["Aggregate rating"].value_counts().sort_index()

print(rating_counts)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=rating_counts.index,
    y=rating_counts.values
)

plt.title("Distribution of Aggregate Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.show()
```

```python
zero_rating_count = (df["Aggregate rating"] == 0).sum()
total_restaurants = len(df)

zero_rating_percentage = (zero_rating_count / total_restaurants) * 100

print("Restaurants with 0.0 rating:", zero_rating_count)
print("Total restaurants:", total_restaurants)
print("Percentage with 0.0 rating:", round(zero_rating_percentage, 2), "%")
```

```python
most_common_rating = df["Aggregate rating"].value_counts().idxmax()
most_common_count = df["Aggregate rating"].value_counts().max()
most_common_percentage = (most_common_count / len(df)) * 100

print("Most common rating:", most_common_rating)
print("Number of restaurants:", most_common_count)
print("Percentage of restaurants:", round(most_common_percentage, 2), "%")
```

## Class Imbalance Observation

The distribution of the `Aggregate rating` variable shows that the ratings are not evenly distributed across all classes.

The rating `0.0` has a noticeably high number of restaurants compared with several other rating categories. This indicates a class imbalance in the target variable.

The most common rating and its percentage were calculated above to quantify this imbalance.

This imbalance should be considered when performing further analysis or building a machine learning model using `Aggregate rating` as the target variable.
