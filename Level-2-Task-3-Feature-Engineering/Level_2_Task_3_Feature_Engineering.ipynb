```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Dataset .csv")

# Display first 5 rows
df.head()
```

```python
# Create new features based on text length

df["Restaurant Name Length"] = df["Restaurant Name"].str.len()

df["Address Length"] = df["Address"].str.len()

# Display the new features
df[[
    "Restaurant Name",
    "Restaurant Name Length",
    "Address",
    "Address Length"
]].head()
```

```python
# Encode categorical variables into numerical features

df["Has Table Booking"] = df["Has Table booking"].map({
    "Yes": 1,
    "No": 0
})

df["Has Online Delivery"] = df["Has Online delivery"].map({
    "Yes": 1,
    "No": 0
})

# Display the encoded features
df[[
    "Has Table booking",
    "Has Table Booking",
    "Has Online delivery",
    "Has Online Delivery"
]].head()
```

```python
# Display all engineered features

engineered_features = df[[
    "Restaurant Name Length",
    "Address Length",
    "Has Table Booking",
    "Has Online Delivery"
]]

print("Engineered Features:")
display(engineered_features.head(10))
```

```python
# Summary statistics of engineered numerical features

print("Summary Statistics:")
display(engineered_features.describe())
```

```python
print("Feature Engineering Summary:")
print("1. Restaurant Name Length was extracted from the Restaurant Name column.")
print("2. Address Length was extracted from the Address column.")
print("3. Has Table Booking was encoded as 1 for Yes and 0 for No.")
print("4. Has Online Delivery was encoded as 1 for Yes and 0 for No.")
print("\nFeature engineering completed successfully.")
```
