```python
country_counts = df["Country Code"].value_counts()

print("Restaurant count by Country Code:")
print(country_counts)
```

```python
city_counts = df["City"].value_counts()

print("Top 10 cities by number of restaurants:")
print(city_counts.head(10))
```

```python
cuisine_counts = df["Cuisines"].value_counts()

print("Top 10 cuisines:")
print(cuisine_counts.head(10))
```

```python
top_cuisines = df["Cuisines"].value_counts().head(10)

print("Top 10 cuisines by number of restaurants:")
print(top_cuisines)
```

```python
plt.figure(figsize=(10, 6))

top_cuisines.sort_values().plot(kind="barh")

plt.title("Top 10 Cuisines by Number of Restaurants")
plt.xlabel("Number of Restaurants")
plt.ylabel("Cuisine")
plt.tight_layout()

plt.show()
```

```python
top_cities = df["City"].value_counts().head(10)

print("Top 10 cities by number of restaurants:")
print(top_cities)
```

```python
plt.figure(figsize=(10, 6))

top_cities.sort_values().plot(kind="barh")

plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("Number of Restaurants")
plt.ylabel("City")
plt.tight_layout()

plt.show()
```

## Findings and Interpretation

### 1. Numerical Analysis
Descriptive statistics were calculated for the numerical variables to understand their central tendency and variability. Measures such as mean, median, standard deviation, minimum, and maximum helped summarize the dataset.

### 2. Categorical Analysis
The categorical variables were analyzed using frequency counts to understand the distribution of restaurants across different cities and cuisines.

### 3. Top Cuisines
North Indian cuisine has the highest number of restaurant entries, followed by North Indian, Chinese and other popular cuisine categories.

### 4. Top Cities
New Delhi has the highest number of restaurants in the dataset, followed by Gurgaon and Noida. These cities have a substantially larger number of restaurant entries compared with the other cities.

### 5. Overall Insight
The analysis shows that restaurants are concentrated in a few major cities, with North Indian cuisine being particularly prominent. Frequency analysis and descriptive statistics provide useful insights into the structure and distribution of the restaurant dataset.
